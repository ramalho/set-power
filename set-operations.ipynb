```python
def primes(stop):
    sieve = [True] * stop
    for n in range(2, stop):
        if sieve[n]:
            yield n
            for multiple in range(n * n, stop, n):
                sieve[multiple] = False
```

```python
prime = set(primes(10))
prime
```

```python
prime & even
```

```python
def fibonacci(stop):
    a, b, = 0, 1
    while a < stop:
        yield a
        a, b, = b, a + b
```

```python
fibo = set(fibonacci(10))
fibo
```

```python
fibo - prime
```

```python
prime - fibo
```
