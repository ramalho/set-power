```python
def fibonacci(stop):
    a, b, = 0, 1
    while a < stop:
        yield a
        a, b, = b, a + b
```

```python
f = set(fibonacci(10))
f
```

```python
def primes(stop):
    sieve = [True] * stop
    for n in range(2, stop):
        if sieve[n]:
            yield n
            for multiple in range(n * n, stop, n):
                sieve[multiple] = False
```

```python
p = set(primes(10))
p
```
